```python
df = pd.read_csv("World Happiness dataset.csv")
```

```python
print(df.head())
print(df.shape)
print(df.columns)
```

```python
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Happiness Score"], kde=True)
plt.title("Distribution of Happiness Score")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.show()
```

```python
top10 = df.sort_values("Happiness Score", ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=top10, x="Happiness Score", y="Country")
plt.title("Top 10 Happiest Countries")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Economy (GDP per Capita)",
    y="Happiness Score"
)

plt.title("GDP vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Family",
    y="Happiness Score"
)

plt.title("Social Support vs Happiness Score")
plt.xlabel("Social Support (Family)")
plt.ylabel("Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Health (Life Expectancy)",
    y="Happiness Score"
)

plt.title("Life Expectancy vs Happiness Score")
plt.xlabel("Healthy Life Expectancy")
plt.ylabel("Happiness Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Freedom",
    y="Happiness Score"
)
plt.title("Freedom vs Happiness Score")
plt.xlabel("Freedom")
plt.ylabel("Happiness Score")
plt.show()
```

```python
factors = [
    "Happiness Score",
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom"
]
corr = df[factors].corr()
print(corr)
```

```python
plt.figure(figsize=(9,6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Between Happiness and Major Factors")
plt.show()
```

```python
happiness_corr = corr["Happiness Score"].sort_values(ascending=False)
print(happiness_corr)
```

```python
sns.pairplot(
    df[
        [
            "Happiness Score",
            "Economy (GDP per Capita)",
            "Family",
            "Health (Life Expectancy)",
            "Freedom"
        ]
    ]
)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df[
        [
            "Happiness Score",
            "Economy (GDP per Capita)",
            "Family",
            "Health (Life Expectancy)",
            "Freedom"
        ]
    ]
)

plt.xticks(rotation=30)
plt.title("Boxplot of Happiness Factors")
plt.show()
```
